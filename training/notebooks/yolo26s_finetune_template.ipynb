```python
from ultralytics import YOLO

DATA_YAML = '/kaggle/input/aerovision-seraphim-subset/data.yaml'
MODEL_WEIGHTS = 'yolo26s.pt'

model = YOLO(MODEL_WEIGHTS)
model.train(data=DATA_YAML, task='detect', imgsz=640, epochs=50, patience=10, seed=42, name='yolo26s-seraphim-subset-v1')
```

Fallback only after YOLO26 unavailability is reported and documented: use `yolo11s.pt` and record `YOLO11` in model card and experiment metadata.
